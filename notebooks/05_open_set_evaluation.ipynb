# 05 - Open-Set Evaluation

Evaluates the confidence-aware hybrid fusion for open-set (withheld attack) detection.

```python
import sys; sys.path.insert(0, '..')
import yaml, pandas as pd

with open('../results/metrics/fusion_results.yaml') as f:
    fusion = yaml.safe_load(f)

print('=== Four Conditions Comparison (Held-out Set) ===')
conditions = ['classifier_only','anomaly_only','simple_fusion','tuned_fusion']
for cond in conditions:
    r = fusion.get(cond, {})
    print(f'\n{cond}:')
    for k, v in r.items():
        if isinstance(v, (int, float)):
            print(f'  {k}: {v:.4f}')

print('\n=== Open-Set Evaluation (Worms Withheld) ===')
openset = fusion.get('openset_worms', {})
for k, v in openset.items():
    print(f'  {k}: {v}')
```

```python
print('=== Calibration ===')
calib = fusion.get('calibration', {})
print(f'ECE:         {calib.get("ece",0):.4f}')
print(f'Brier Score: {calib.get("brier_score",0):.4f}')
print(f'Method:      {calib.get("method","N/A")}')

print('\n=== Operational Metrics ===')
op = fusion.get('operational', {})
for k, v in op.items():
    print(f'  {k}: {v}')
```
