# 02 - Preprocessing Pipeline

This notebook demonstrates the preprocessing pipeline on UNSW-NB15 data.
All transformations are fitted on training data only.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
from src.preprocessing import NIDSPreprocessor
from src.data_loader import load_raw_dataset

df_train = load_raw_dataset('training')
df_test  = load_raw_dataset('testing')
print(f'Train: {df_train.shape}, Test: {df_test.shape}')
```

```python
# Fit on training only
preprocessor = NIDSPreprocessor()
X_train = preprocessor.fit_transform(df_train)
X_test  = preprocessor.transform(df_test)
y_train = preprocessor.get_labels(df_train, 'multiclass')
y_test  = preprocessor.get_labels(df_test, 'multiclass')
print(f'X_train: {X_train.shape}, X_test: {X_test.shape}')
print(f'Features: {preprocessor.feature_names}')
print(f'Label classes: {preprocessor.label_encoder.classes_.tolist()}')
```

```python
# Verify no data leakage: test scaler was not refit
print('X_train mean ~0:', abs(X_train.mean()) < 0.01)
print('X_train std ~1:', abs(X_train.std() - 1.0) < 0.1)
print('No NaN in train:', not np.isnan(X_train).any())
print('No NaN in test:', not np.isnan(X_test).any())
```
