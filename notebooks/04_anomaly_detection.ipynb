# 04 - Anomaly Detection

Evaluates Isolation Forest trained on normal-only traffic.

```python
import sys; sys.path.insert(0, '..')
import yaml, matplotlib.pyplot as plt, numpy as np

with open('../results/metrics/anomaly_results.yaml') as f:
    anom = yaml.safe_load(f)

print(f'AUROC:          {anom.get("auroc",0):.4f}')
print(f'AUPRC:          {anom.get("auprc",0):.4f}')
print(f'Precision:      {anom.get("precision",0):.4f}')
print(f'Recall:         {anom.get("recall",0):.4f}')
print(f'F1:             {anom.get("f1",0):.4f}')
print(f'FPR on Normal:  {anom.get("fpr_on_normal",0):.4f}')
print(f'Threshold:      {anom.get("threshold",0):.4f}')
```

```python
per_class = anom.get('per_class_detection_rates', {})
if per_class:
    plt.figure(figsize=(10,5))
    plt.bar(per_class.keys(), per_class.values(), color='#e74c3c')
    plt.xlabel('Attack Category')
    plt.ylabel('Detection Rate')
    plt.title('Isolation Forest Detection Rate per Attack Category')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('../results/figures/anomaly_detection_rates.png', dpi=150, bbox_inches='tight')
    plt.show()
```
