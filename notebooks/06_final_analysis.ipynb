# 06 - Final Analysis

Consolidated analysis of all experiment results for research paper preparation.

```python
import sys; sys.path.insert(0, '..')
import yaml, pandas as pd, numpy as np, matplotlib.pyplot as plt

with open('../results/metrics/classifier_results.yaml') as f: clf = yaml.safe_load(f)
with open('../results/metrics/anomaly_results.yaml') as f: anom = yaml.safe_load(f)
with open('../results/metrics/fusion_results.yaml') as f: fusion = yaml.safe_load(f)

print('=' * 60)
print('COMPLETE RESULTS SUMMARY')
print('=' * 60)
print('\n--- Phase 2: Multiclass Classification ---')
xgb = clf.get('xgboost', {})
rf = clf.get('random_forest', {})
summary = pd.DataFrame({'Model': ['XGBoost','Random Forest'],
    'Accuracy': [xgb.get('accuracy',0), rf.get('accuracy',0)],
    'Macro-F1': [xgb.get('macro_f1',0), rf.get('macro_f1',0)],
    'Weighted-F1': [xgb.get('weighted_f1',0), rf.get('weighted_f1',0)]})
print(summary.to_string(index=False))

print('\n--- Phase 3: Anomaly Detection ---')
print(f'  AUROC: {anom.get("auroc",0):.4f} | AUPRC: {anom.get("auprc",0):.4f}')

print('\n--- Phase 4: Fusion (Open-Set Worms) ---')
openset = fusion.get('openset_worms', {})
print(f'  Worms Recall: {openset.get("unknown_recall",0):.4f}')
print(f'  FPR: {openset.get("false_positive_rate",0):.4f}')

print('\n--- Phase 5: Top SHAP Features ---')
import os
ranking_path = '../results/metrics/shap_feature_ranking.txt'
if os.path.exists(ranking_path):
    with open(ranking_path) as f: print(f.read())
```
