# 03 - Multiclass Models

Trains and evaluates Random Forest and XGBoost on UNSW-NB15.
See `experiments/train_classifiers.py` for the full training script.

```python
import sys; sys.path.insert(0, '..')
import yaml, pandas as pd

with open('../results/metrics/classifier_results.yaml') as f:
    results = yaml.safe_load(f)

print('=== XGBoost ===')
xgb = results.get('xgboost', {})
print(f'Accuracy:    {xgb.get("accuracy", 0):.4f}')
print(f'Macro-F1:    {xgb.get("macro_f1", 0):.4f}')
print(f'Weighted-F1: {xgb.get("weighted_f1", 0):.4f}')

print('\n=== Random Forest ===')
rf = results.get('random_forest', {})
print(f'Accuracy:    {rf.get("accuracy", 0):.4f}')
print(f'Macro-F1:    {rf.get("macro_f1", 0):.4f}')
print(f'Weighted-F1: {rf.get("weighted_f1", 0):.4f}')
```

```python
import matplotlib.pyplot as plt, numpy as np
import pandas as pd

per_class = xgb.get('per_class_report', {})
if per_class:
    classes = list(per_class.keys())
    f1_scores = [per_class[c].get('f1', 0) for c in classes]
    plt.figure(figsize=(12, 5))
    bars = plt.barh(classes, f1_scores, color=['#2ecc71' if f > 0.7 else '#e74c3c' for f in f1_scores])
    plt.xlabel('F1-Score')
    plt.title('XGBoost Per-Class F1 (Test Set)')
    plt.axvline(x=np.mean(f1_scores), color='orange', linestyle='--', label=f'Macro-F1={np.mean(f1_scores):.3f}')
    plt.legend()
    plt.tight_layout()
    plt.savefig('../results/figures/xgboost_per_class_f1.png', dpi=150, bbox_inches='tight')
    plt.show()
```
